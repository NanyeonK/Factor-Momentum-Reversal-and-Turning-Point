```python
import pandas as pd

# 데이터 파일 경로
file_path = "C:/Users/YeonChan Kang/Downloads/F-F_Research_Data_Factors_CSV/F-F_Research_Data_Factors.CSV"

# CSV 파일 읽기
df = pd.read_csv(file_path, skiprows=3)

# 컬럼 이름 변경
df.columns = ['Date', 'Mkt-RF', 'SMB', 'HML', 'RF']

# 날짜 형식 변환 및 인덱스 설정
df['Date'] = pd.to_datetime(df['Date'], format='%Y%m').dt.to_period('M')
df.set_index('Date', inplace=True)

# 첫 번째 행이 NA인 경우 해당 행 삭제
df.dropna(inplace=True)

# 데이터 출력 (첫 몇 행만)
print(df.head())
```
