```python
def newey_west_regression(y, X, lags=6):
    """
    Performs a regression and calculates Newey-West standard errors, t-statistics, and p-values.
    
    Args:
        y (pd.Series or np.array): Dependent variable.
        X (pd.DataFrame or np.array): Independent variables.
        lags (int): The lag length for Newey-West standard errors. If None, it will be set to the integer part of the cube root of the number of observations.
    
    Returns:
        pd.DataFrame: A DataFrame containing coefficients, Newey-West standard errors, t-statistics, and p-values.
    """
    if lags is None:
        lags = int(np.floor(len(y)**(1/3)))  # Set lag length to the integer part of the cube root of the number of observations

    # Ensure the dependent variable is numeric
    y = pd.to_numeric(y, errors='coerce')
    
    # Ensure all independent variables are numeric
    X = X.apply(pd.to_numeric, errors='coerce')

    # Drop rows with any NaN values in y or X
    valid_idx = ~y.isna() & X.notna().all(axis=1)
    y = y[valid_idx]
    X = X[valid_idx]
    
    # Add a constant term to the independent variables matrix
    X = sm.add_constant(X)
    
    # Estimate the regression model
    model = sm.OLS(y, X).fit()
    
    # Calculate Newey-West standard errors
    robust_cov = model.get_robustcov_results(cov_type='HAC', maxlags=lags)
    
    # Extract coefficients, standard errors, t-statistics, and p-values
    results = pd.DataFrame({
        'Coefficient': model.params,
        'Newey-West SE': robust_cov.bse,
        't-Statistic': robust_cov.tvalues,
        'p-Value': robust_cov.pvalues
    })
    
    return results
```

```python
from scipy.stats import pearsonr, spearmanr

def cal_corr(group, var1, var2, option='all'):
    """
    Calculate the correlation between two variables for a given period.

    Args:
        group (pd.DataFrame): The data for the given period.
        var1 (str): The name of the first variable.
        var2 (str): The name of the second variable.
        option (str): The correlation type to calculate ('pearson', 'spearman', or 'all'). Defaults to 'all'.
        
    Returns:
        pd.Series: A series containing the correlation between the two variables.
    """
    # Drop NaN values for the variables being correlated
    group = group[[var1, var2]].dropna()
    
    if len(group) < 2:
        return pd.Series({'Pearson': np.nan, 'Spearman': np.nan})

    if option == 'pearson':
        pearson_corr, _ = pearsonr(group[var1], group[var2])
        return pd.Series({'Pearson': pearson_corr})
    elif option == 'spearman':
        spearman_corr, _ = spearmanr(group[var1], group[var2])
        return pd.Series({'Spearman': spearman_corr})
    elif option == 'all':
        pearson_corr, _ = pearsonr(group[var1], group[var2])
        spearman_corr, _ = spearmanr(group[var1], group[var2])
        return pd.Series({'Pearson': pearson_corr, 'Spearman': spearman_corr})
    else:
        raise ValueError("Invalid option for correlation type. Choose from 'pearson', 'spearman', or 'all'.")

def cal_per_corr(df, time_column, specific_date=None):
    """
    Calculate Pearson and Spearman correlations for each time period for all pairs of variables.
    
    Args:
        df (pd.DataFrame): The data frame containing the data.
        time_column (str): The name of the column representing time periods.
        specific_date (str or None): A specific date to filter the data. If None, calculate for all dates.
    
    Returns:
        pd.DataFrame: A data frame containing the Pearson and Spearman correlations for each time period.
    """
    if specific_date:
        df = df[df[time_column] == specific_date]

    variables = [col for col in df.columns if col != time_column]
    correlations = []

    for i, var1 in enumerate(variables):
        for var2 in variables[i+1:]:
            corr_df = df.groupby(time_column).apply(lambda group: cal_corr(group, var1, var2)).reset_index()
            corr_df['Var1'] = var1
            corr_df['Var2'] = var2
            correlations.append(corr_df)

    all_correlations = pd.concat(correlations, ignore_index=True)
    return all_correlations
```

```python
def cal_ts_avcorr(correlations_df):
    """
    Calculate the time-series averages of the periodic cross-sectional correlations.
    
    Args:
        correlations_df (pd.DataFrame): A data frame containing the periodic correlations.
    
    Returns:
        pd.DataFrame: A data frame containing the time-series average Pearson and Spearman correlations.
    """
    avg_corrs = correlations_df.groupby(['Var1', 'Var2'])[['Pearson', 'Spearman']].mean().reset_index()
    return avg_corrs
```

```python
def create_corr_mat(avg_corrs, variables):
    """
    Create a correlation matrix with Pearson correlations below the diagonal and Spearman correlations above the diagonal.
    
    Args:
        avg_corrs (pd.DataFrame): A data frame containing the average correlations.
        variables (list of str): List of column names representing the variables.
    
    Returns:
        pd.DataFrame: A correlation matrix.
    """
    matrix = pd.DataFrame(index=variables, columns=variables)

    for _, row in avg_corrs.iterrows():
        var1, var2 = row['Var1'], row['Var2']
        pearson, spearman = row['Pearson'], row['Spearman']
        matrix.at[var1, var2] = f"{spearman:.2f}"
        matrix.at[var2, var1] = f"{pearson:.2f}"
    
    np.fill_diagonal(matrix.values, np.nan)
    return matrix
```

```python
class UnivariatePortfolioAnalyzer:
    def __init__(self, df, time_column, id_column):
        """
        Initialize the UnivariatePortfolioAnalyzer with the data frame, time column, and ID column.
        
        Args:
            df (pd.DataFrame): The data frame containing the data.
            time_column (str): The name of the column representing time periods.
            id_column (str): The name of the column representing unique entity IDs.
        """
        self.df = df
        self.time_column = time_column
        self.id_column = id_column
    
    def cal_bp(self, value_column, num_portfolios, custom_percentiles=None):
        """
        Calculate breakpoints for a given variable based on specified quantiles for univariate portfolio analysis.
        
        Args:
            value_column (str): The name of the column representing the values to calculate breakpoints for.
            num_portfolios (int): The number of portfolios to be formed each time period.
            custom_percentiles (list of float, optional): Custom percentiles to calculate breakpoints. If None, evenly spaced percentiles are used.
        
        Returns:
            pd.DataFrame: A data frame containing the breakpoints for each time period.
        """
        if custom_percentiles:
            assert all(0 < p < 100 for p in custom_percentiles), "Percentiles must be between 0 and 100"
            percentiles = custom_percentiles
        else:
            percentiles = [k * 100 / num_portfolios for k in range(1, num_portfolios)]
        
        quantiles = [p / 100 for p in percentiles]
        breakpoints = self.df.groupby(self.time_column)[value_column].quantile(quantiles).unstack()
        breakpoints.columns = [f'B{k+1} ({round(percentiles[k], 3)})' for k in range(len(percentiles))]
        
        return breakpoints

    def cal_multi_bp(self, characteristics, num_portfolios=5, custom_percentiles=None):
        """
        Calculate breakpoints for multiple characteristics in the dataset using the cal_bp function.

        Parameters:
        - characteristics (list of str): List of column names of the firm characteristics to calculate breakpoints for.
        - num_portfolios (int): The number of portfolios to calculate. Default is 5.
        - custom_percentiles (list of float): Custom percentiles for breakpoints. Should be between 0 and 100.

        Returns:
        - breakpoints_dict (dict): A dictionary where keys are characteristic names and values are DataFrames with dates as index and breakpoints as columns.
        """
        breakpoints_dict = {}
        
        for characteristic in characteristics:
            breakpoints_dict[characteristic] = self.cal_bp(characteristic, num_portfolios, custom_percentiles)

        return breakpoints_dict

    def assign_portfolios(self, breakpoints, value_column, keep_columns=None, method='drop'):
        """
        Assign portfolios based on calculated breakpoints.

        Args:
            breakpoints (pd.DataFrame): The data frame containing the breakpoints for each time period.
            value_column (str): The name of the column representing the values to assign to portfolios.
            keep_columns (list of str, optional): List of column names to keep in the resulting DataFrame. If None, only time_column, value_column, and portfolio are kept.
            method (str): Method to handle duplicate breakpoints. 'drop' to drop duplicates, 'modify' to add small random values.
        
        Returns:
            pd.DataFrame: The data frame with the specified columns and an additional column for portfolio assignment.
        """
        if keep_columns is None:
            keep_columns = []

        df = self.df[[self.id_column, self.time_column, value_column] + keep_columns].copy()

        # Remove rows with NaN values in the value column
        df = df.dropna(subset=[value_column])

        # Check for and handle duplicate index labels
        if df.index.duplicated().any():
            df = df.reset_index(drop=True)
        
        df['portfolio'] = np.nan

        for time_period in breakpoints.index:
            period_data = df[df[self.time_column] == time_period]
            period_breakpoints = breakpoints.loc[time_period].dropna().values
            
            if method == 'drop':
                period_breakpoints = np.unique(period_breakpoints)
            elif method == 'modify':
                period_breakpoints += np.random.uniform(0, 1e-10, size=period_breakpoints.shape)
            
            # Ensure breakpoints are monotonically increasing
            period_breakpoints = np.sort(period_breakpoints)
            if not np.all(np.diff(period_breakpoints) > 0):
                print(f"Warning: Non-monotonic breakpoints detected for period {time_period}.")
                continue
            
            portfolios = pd.cut(
                period_data[value_column],
                bins=[-np.inf] + period_breakpoints.tolist() + [np.inf],
                labels=[f'P{k+1}' for k in range(len(period_breakpoints) + 1)],
                include_lowest=True
            )
            df.loc[period_data.index, 'portfolio'] = portfolios

        return df[[self.id_column, self.time_column, value_column] + keep_columns + ['portfolio']]

    def assign_multi_portfolios(self, breakpoints_dict, keep_columns=None, method='drop'):
        """
        Assign portfolios for multiple characteristics based on their respective breakpoints.

        Args:
            breakpoints_dict (dict): A dictionary containing the breakpoints for each characteristic.
            keep_columns (list of str, optional): List of column names to keep in the resulting DataFrame.
            method (str): Method to handle duplicate breakpoints. 'drop' to drop duplicates, 'modify' to add small random values.
        
        Returns:
            dict: A dictionary where keys are characteristic names and values are DataFrames with portfolio assignments.
        """
        if keep_columns is None:
            keep_columns = []

        portfolios_dict = {}
        for characteristic, breakpoints in breakpoints_dict.items():
            portfolios = self.assign_portfolios(breakpoints, characteristic, keep_columns=keep_columns, method=method)
            portfolios_dict[characteristic] = portfolios
        
        return portfolios_dict
    
    def cal_num_per_port(self, portfolio_column):
        """
        Calculate the number of stocks in each portfolio per time period.

        Args:
            portfolio_column (str): The name of the column representing portfolio assignments.
        
        Returns:
            pd.DataFrame: A data frame containing the number of stocks in each portfolio for each time period.
        """
        return self.df.groupby([self.time_column, portfolio_column]).size().unstack(fill_value=0)
    
    def cal_multi_num_per_port(self, portfolio_df_dict):
        results = {}
        for col, df in portfolio_df_dict.items():
            results[col] = df.groupby([self.time_column, f'{col}_portfolio']).size().unstack(fill_value=0)
        return results
    
    def cal_avg_port_values(self, portfolio_df, portfolio_col, target_col, weight_type=None):
        """
        Calculate the average values for each portfolio and the difference between the highest and lowest portfolios.
        
        Args:
            portfolio_df (pd.DataFrame): The data frame containing the portfolio data.
            portfolio_col (str): The name of the column representing portfolio assignments.
            target_col (str): The name of the column representing the values or returns.
            weight_type (str, optional): The type of weighting ('value' for market_equity or None for equal weights).
        
        Returns:
            pd.DataFrame: A data frame containing the average values for each portfolio and the difference.
        """
        if weight_type is None:
            portfolio_df['weight'] = 1
        elif weight_type == 'value':
            portfolio_df['weight'] = portfolio_df['market_equity'] / portfolio_df.groupby([self.time_column, portfolio_col])['market_equity'].transform('sum')
        else:
            raise ValueError("Invalid weight type. Choose from None or 'value'.")

        def weighted_avg(group):
            values = group[target_col].values
            weights = group['weight'].values
            return np.dot(values, weights) / weights.sum()

        weighted_values = portfolio_df.groupby([self.time_column, portfolio_col]).apply(weighted_avg).unstack(fill_value=0)

        high_portfolio = weighted_values.columns[-1]
        low_portfolio = weighted_values.columns[0]
        weighted_values['diff'] = weighted_values[high_portfolio] - weighted_values[low_portfolio]

        return weighted_values.reset_index()

    def cal_multi_avg_port_values(self, portfolio_df_dict, target_column, weight_column=None):
        results = {}
        for col, df in portfolio_df_dict.items():
            results[col] = self.cal_avg_port_values(df, col, target_column, weight_column)
        return results
    
    @staticmethod
    def newey_west_regression(y, X, lags=None):
        if lags is None:
            lags = int(np.floor(len(y) ** (1 / 3)))

        y = pd.to_numeric(y, errors='coerce')
        X = X.apply(pd.to_numeric, errors='coerce')
        valid_idx = ~y.isna() & X.notna().all(axis=1)
        y = y[valid_idx]
        X = X[valid_idx]
        X = sm.add_constant(X)
        model = sm.OLS(y, X).fit()
        robust_cov = model.get_robustcov_results(cov_type='HAC', maxlags=lags)
        results = pd.DataFrame({
            'Coefficient': model.params,
            'Newey-West SE': robust_cov.bse,
            't-Statistic': robust_cov.tvalues,
            'p-Value': robust_cov.pvalues
        })
        return results
    
    def summarize_variable(self, avg_values, round_digits=2):
        summary_stats = {}
        for col in avg_values.columns:
            if col != self.time_column:
                y = avg_values[col]
                X = pd.DataFrame(np.ones(len(y)), columns=['Intercept'])
                nw_result = self.newey_west_regression(y, X)
                summary_stats[col] = {
                    'mean': round(y.mean(), round_digits),
                    'Newey-West SE': round(nw_result.loc['Intercept', 'Newey-West SE'], round_digits),
                    't-Statistic': round(nw_result.loc['Intercept', 't-Statistic'], round_digits),
                    'p-Value': round(nw_result.loc['Intercept', 'p-Value'], round_digits)
                }
        
        summary_df = pd.DataFrame(summary_stats)
        return summary_df
    
    def multi_summarize_results(self, avg_values_dict):
        summary_results = {}
        for key, avg_values in avg_values_dict.items():
            summary_results[key] = self.summarize_variable(avg_values)
        return summary_results
    
    def average_return_sorted(self, portfolio_df, portfolio_column):
        """
        Calculate the average returns for each portfolio and the difference between the highest and lowest portfolios.

        Args:
            portfolio_df (pd.DataFrame): The data frame containing the portfolio data.
            portfolio_column (str): The name of the column representing portfolio assignments.

        Returns:
            pd.DataFrame: A data frame with average returns for each portfolio and the difference.
        """
        avg_returns = portfolio_df.groupby(portfolio_column)['ret'].mean().to_frame(name='average_return')
        return avg_returns

    def cal_bench(self, portfolio_df_dict, factors_df, models=['CAPM', 'FF3', 'FF3M', 'FF5', 'FF5M']):
        """
        Calculate Jensen's Alpha for each portfolio based on different benchmark models.

        Args:
            portfolio_df_dict (dict): A dictionary where keys are characteristic names and values are DataFrames with portfolio assignments.
            factors_df (pd.DataFrame): A data frame containing the factor returns for the benchmark models.
            models (list of str): List of benchmark models to use. Defaults to ['CAPM', 'FF3', 'FF3M', 'FF5', 'FF5M'].

        Returns:
            dict: A dictionary where keys are characteristic names and values are DataFrames with Jensen's Alpha and betas for each model.
        """
        results = {}
        for characteristic, portfolio_df in portfolio_df_dict.items():
            portfolio_df = portfolio_df.merge(factors_df, on=self.time_column)
            alphas = {}
            for model in models:
                if model == 'CAPM':
                    X = portfolio_df[['Mkt-RF']]
                elif model == 'FF3':
                    X = portfolio_df[['Mkt-RF', 'SMB', 'HML']]
                elif model == 'FF3M':
                    X = portfolio_df[['Mkt-RF', 'SMB', 'HML', 'UMD']]
                elif model == 'FF5':
                    X = portfolio_df[['Mkt-RF', 'SMB', 'HML', 'RMW', 'CMA']]
                elif model == 'FF5M':
                    X = portfolio_df[['Mkt-RF', 'SMB', 'HML', 'RMW', 'CMA', 'UMD']]
                else:
                    raise ValueError("Invalid model specified.")
                
                portfolio_df['ret_excess'] = portfolio_df['ret'] - portfolio_df['RF']
                y = portfolio_df['ret_excess']
                nw_result = self.newey_west_regression(y, X)
                alphas[model] = nw_result['Coefficient'].to_dict()
            
            results[characteristic] = pd.DataFrame(alphas)
        
        return results
```

```python
class IndependentBivariatePortfolioAnalyzer(UnivariatePortfolioAnalyzer):
    def __init__(self, df, time_column, id_column):
        super().__init__(df, time_column, id_column)
    
    def cal_independent_bp(self, value_column_1, value_column_2, num_portfolios_1, num_portfolios_2, custom_percentiles_1=None, custom_percentiles_2=None):
        breakpoints_1 = self.cal_bp(value_column_1, num_portfolios_1, custom_percentiles_1)
        breakpoints_2 = self.cal_bp(value_column_2, num_portfolios_2, custom_percentiles_2)
        
        breakpoints_dict = {
            value_column_1: breakpoints_1,
            value_column_2: breakpoints_2
        }
        
        return breakpoints_dict

    def assign_independent_portfolios(self, breakpoints_dict, value_column_1, value_column_2, keep_columns=None, method='drop'):
        if keep_columns is None:
            keep_columns = []

        portfolios_dict = {}

        portfolios_1 = self.assign_portfolios(breakpoints_dict[value_column_1], value_column_1, keep_columns=keep_columns, method=method)
        portfolios_1 = portfolios_1[[self.id_column, self.time_column, 'portfolio']].rename(columns={'portfolio': f'{value_column_1}_portfolio'})

        portfolios_2 = self.assign_portfolios(breakpoints_dict[value_column_2], value_column_2, keep_columns=keep_columns, method=method)
        portfolios_2 = portfolios_2[[self.id_column, self.time_column, 'portfolio']].rename(columns={'portfolio': f'{value_column_2}_portfolio'})

        combined_portfolios = pd.merge(portfolios_1, portfolios_2, on=[self.id_column, self.time_column], how='left')

        combined_portfolios['combined_portfolio'] = combined_portfolios.apply(
            lambda row: f"P{row[f'{value_column_1}_portfolio']}_{row[f'{value_column_2}_portfolio']}", axis=1
        )

        portfolios_dict[value_column_1] = portfolios_1
        portfolios_dict[value_column_2] = portfolios_2
        portfolios_dict['combined'] = combined_portfolios

        return portfolios_dict
    
    def cal_avg_portfolio_values(self, portfolios_dict, outcome_variable, weight_column=None):
        combined_portfolios = portfolios_dict['combined'].copy()
        
        if weight_column is None:
            combined_portfolios['weight'] = 1
        else:
            combined_portfolios['weight'] = combined_portfolios[weight_column]

        def weighted_avg(group):
            values = group[outcome_variable].values
            weights = group['weight'].values
            return np.dot(values, weights) / weights.sum()

        avg_values = combined_portfolios.groupby([self.time_column, 'combined_portfolio']).apply(weighted_avg).unstack(fill_value=0)

        portfolios_1 = portfolios_dict[next(iter(portfolios_dict))].drop(columns=[f'{next(iter(portfolios_dict))}_portfolio']).drop_duplicates()
        portfolios_2 = portfolios_dict[next(iter(portfolios_dict))].drop(columns=[f'{next(iter(portfolios_dict))}_portfolio']).drop_duplicates()

        for portfolio_1 in portfolios_1[f'{next(iter(portfolios_dict))}_portfolio'].unique():
            avg_values[f'{portfolio_1}_diff'] = (
                avg_values.loc[:, f'P{portfolio_1}_P{portfolios_2[f"{next(iter(portfolios_dict))}_portfolio"].max()}'] -
                avg_values.loc[:, f'P{portfolio_1}_P{portfolios_2[f"{next(iter(portfolios_dict))}_portfolio"].min()}']
            )
            avg_values[f'{portfolio_1}_avg'] = avg_values.loc[:, [col for col in avg_values.columns if f'P{portfolio_1}_' in col]].mean(axis=1)

        for portfolio_2 in portfolios_2[f'{next(iter(portfolios_dict))}_portfolio'].unique():
            avg_values[f'{portfolio_2}_diff'] = (
                avg_values.loc[:, f'P{portfolios_1[f"{next(iter(portfolios_dict))}_portfolio"].max()}_P{portfolio_2}'] -
                avg_values.loc[:, f'P{portfolios_1[f"{next(iter(portfolios_dict))}_portfolio"].min()}_P{portfolio_2}']
            )
            avg_values[f'{portfolio_2}_avg'] = avg_values.loc[:, [col for col in avg_values.columns if f'_P{portfolio_2}' in col]].mean(axis=1)

        avg_values['overall_avg'] = avg_values.mean(axis=1)
        avg_values['diff_of_diff'] = avg_values[[col for col in avg_values.columns if '_diff' in col]].mean(axis=1)

        return avg_values.reset_index()
```
